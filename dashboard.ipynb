```python
import pandas as pd
import plotly.express as px
from dash import Dash, html, dcc
import dash_bootstrap_components as dbc

# Load dataset
df = pd.read_csv("heart_updated.csv")

# App initialization
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Graphs
bar_chart = px.bar(df, x='Sex', color='HeartDisease',
                   title='Heart Disease by Gender', text_auto=True,
                   template='plotly_white')

hist_chart = px.histogram(df, x='Age', color='HeartDisease',
                          title='Age Distribution by Heart Disease',
                          template='plotly_white')

violin_chart = px.violin(df, x='HeartDisease', y='Cholesterol', color='HeartDisease',
                         box=True, points='all',
                         title='Cholesterol Spread by Heart Disease',
                         template='plotly_white')

scatter_chart = px.scatter(df, x='Cholesterol', y='Age', color='HeartDisease',
                           title='Cholesterol vs Age by Heart Disease',
                           template='plotly_white')

pie_chart = px.pie(df, names='Sex', title='Gender Distribution',
                   template='plotly_white')

# Layout
app.layout = dbc.Container([
    html.H1("Heart Disease Dashboard", className="text-center my-4"),

    dbc.Row([
        dbc.Col(dcc.Graph(figure=bar_chart), md=6),
        dbc.Col(dcc.Graph(figure=hist_chart), md=6)
    ]),

    dbc.Row([
        dbc.Col(dcc.Graph(figure=violin_chart), md=6),
        dbc.Col(dcc.Graph(figure=scatter_chart), md=6)
    ]),

    dbc.Row([
        dbc.Col(dcc.Graph(figure=pie_chart), md=6)
    ]),
], fluid=True)

# Run
if __name__ == "__main__":
    app.run(debug=True)

```
