```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("heart.csv")


fig = px.histogram(df, x='Sex', color='HeartDisease', barmode='group', title='Heart Disease Distribution by Gender')
fig.show()
```

```python
fig = px.box(df, x='HeartDisease', y='Age', color='HeartDisease', title='Age Distribution by Heart Disease')
fig.show()
```

```python
fig = px.scatter(df, x='Age', y='MaxHR', color='HeartDisease', trendline='ols',
                 title='Age vs MaxHR with Heart Disease')
fig.show()
```

chest pain type and heart disease relation 

```python
fig = px.histogram(df, x='ChestPainType', color='HeartDisease', barmode='group',
                   title='Heart Disease Distribution by Chest Pain Type')
fig.show()
```

resting blood pressure

```python
fig = px.violin(df, y='RestingBP', x='HeartDisease', box=True, points='all',
                color='HeartDisease', title='Resting Blood Pressure by Heart Disease')
fig.show()
```

cholesterol relation with heart disease

```python
fig = px.scatter(df, x='Cholesterol', y='Age', color='HeartDisease',
                 title='Cholesterol vs Age by Heart Disease')
fig.show()
```

```python
fig = px.treemap(df, path=['Sex', 'ChestPainType'], values='HeartDisease',
                 color='HeartDisease', 
                 title='Heart Disease Cases by Gender and Chest Pain Type',
                 color_continuous_scale='Reds')
fig.show()
```

```python
fig = px.histogram(df, x='ChestPainType', color='HeartDisease',
                   facet_col='Sex',
                   barmode='stack',
                   title='Heart Disease vs Chest Pain Type by Gender')
fig.show()

```
